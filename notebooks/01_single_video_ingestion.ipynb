# 01 - Single-video TikTok ingestion

Ingest one public TikTok URL into immutable raw artifacts under `data/raw/<video_id>/`, then inspect the normalized metadata, manifest, and MP4.

Set `VIDEO_URL` below to any public TikTok video URL.

```python
from tiktok_analytics_factory.ingestion import ingest

VIDEO_URL = "https://www.tiktok.com/@scout2015/video/6718335390845095173"  # example; replace freely
OUTPUT_ROOT = "data/raw"

result = ingest(VIDEO_URL, OUTPUT_ROOT)
print("video id:", result.video_id)
print("artifacts:", result.artifact_dir)
```

```python
import json

print(json.dumps(result.metadata.to_json_dict(), indent=2))
```

```python
print(json.dumps(result.manifest, indent=2))
```
